## Data Ingestion

### Langchain  Document Structure :- 

It is kind of dataStructure wich will be able to save some data in fomat like page content(actual content wich available in file ) and meta data(additional info about file like file name , time , type of file )
  
in langchain we have diff diff Loader like:- pdf , csv , webbase,directory etc
it used to load the pdf file after loading it will show the o/p in form of document structure








```python
## Document Structure


from langchain_core.documents import Document

```

```python
doc=Document(
    page_content="this is simple content for an example used",
    metadata={
        "file_name":"example_txt",
        "author":"pruthviraj Thorbole",
        "no_page":1,
        "date_created":"01-01-2026"

    }
)
doc
```

```python
### Text Loader 

from langchain_community.document_loaders import TextLoader

loader=TextLoader("../data/text_files/python_intro.txt",encoding="utf-8")

document=loader.load()

print(document)
```

```python
### Directory Loader

from langchain_community.document_loaders import DirectoryLoader
from tqdm import tqdm

dir_loader=DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding":"utf-8"},
    show_progress=True

)

documents=dir_loader.load()

print(documents)
```

```python
### PDF LOADER

from langchain_community.document_loaders import PyMuPDFLoader


pdf_loader=DirectoryLoader(
    "../data/pdf",
    glob="**/*.pdf",
    loader_cls=PyMuPDFLoader,
    show_progress=True


)

pdf_document=pdf_loader.load()

print(pdf_document)
```
